```python
import pandas as pd
```

```python
df=pd.read_csv('/content/globalterrorismdb_0718dist.csv (2).zip',encoding="latin1")
df.head()
```

```python
df.isnull().sum()
```

```python
(df.isnull().mean() * 100).sort_values(ascending=False)
```

```python
threshold = len(df) * 0.5
df = df.dropna(axis=1, thresh=threshold)
```

```python
df = df.dropna(subset=['iyear', 'country_txt'])
```

```python
df['nkill'] = df['nkill'].fillna(0)
df['nwound'] = df['nwound'].fillna(0)
```

```python
df['city'] = df['city'].fillna('Unknown')
```

```python
df.info()
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
cols_to_drop = [
    'nperps', 'nperpcap',
    'weapdetail',
    'scite1',
    'summary','weapsubtype1','targsubtype1','natlty1']

df = df.drop(columns=cols_to_drop)
```

```python
fill_zero_cols = [
    'nkillter', 'nkillus',
    'nwoundte', 'nwoundus','claimed','guncertain1','ishostkid']

df[fill_zero_cols] = df[fill_zero_cols].fillna(0)
```

```python
fill_unknown_cols=['corp1','weapsubtype1_txt','targsubtype1_txt','natlty1_txt','target1','provstate']
df[fill_unknown_cols]=df[fill_unknown_cols].fillna('Unknown')
```

```python
cols_to_drop = [
    'INT_LOG', 'INT_IDEO',
    'INT_MISC',
    'INT_ANY']

df = df.drop(columns=cols_to_drop)
```

```python
df_map = df.dropna(subset=['latitude', 'longitude'])
```

```python
df = df.dropna(subset=['specificity', 'doubtterr', 'multiple'])
```
